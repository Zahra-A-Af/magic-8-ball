```python
import random

def magic_8_ball():
    #answers
    affirmative = ["It is certain.", "Without a doubt.", "Yes, definitely."]
    non_committal = ["Reply hazy, try again.", "Ask again later.", "Cannot predict now."]
    negative = ["Don't count on it.", "My reply is no.", "Very doubtful."]
    all_answers = affirmative + non_committal + negative

    print("Welcome to the Magic 8 Ball!")
    #The loop for continuing
    while True:
        question = input("\nAsk your yes/no question: ")
        answer = random.choice(all_answers)
        print("The Magic 8 Ball says:", answer)

        again = input("\nDo you want to ask another question? (yes/no): ").strip().lower()

        if again != 'yes':
            print("Goodbye! 👋")
            break
```

```python
magic_8_ball()
```
